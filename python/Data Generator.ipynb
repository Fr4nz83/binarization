```python
import numpy as np
```

```python
size_matrix = 256

batch = size_matrix
num_feats = size_matrix
num_hidden = size_matrix
```

```python
d = (np.random.rand(batch, num_feats) - 0.5) * 10
d = d.astype(np.float32)
display(d.shape, d.dtype, d)
```

```python
w = (np.random.rand(num_feats, num_hidden) - 0.5) * 10
w = w.astype(np.float32)
display(w.shape, w.dtype, w)
```

```python
b = (np.random.rand(num_hidden) - 0.5) * 10
b = b.astype(np.float32)
display(b.shape, b.dtype, b)
```

```python
np.save("d.npy", d)
np.save("w.npy", w)

# Importante salvare esplicitamente la trasposta in row_major (order=C)
np.save("fc1.weight.npy", np.copy(w.T, order='C'))

np.save("b.npy", b)
np.save("fc1.bias.npy", b)
```
