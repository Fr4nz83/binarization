```python
import numpy as np
```

#### Load dataset

```python
dataset = np.load("d.npy")

display(dataset)
display(dataset.shape)

dataset = np.where(dataset >= 0, 1, -1)
display(dataset)
```

#### Load weights layers.

```python
w = np.load("w.npy")

display(w, w.shape)
display(w.T, w.T.shape)

w = np.where(w >= 0, 1, -1)
display(w, w.shape)
display(w.T, w.T.shape)
```

#### Load bias layers.

```python
b = np.load("b.npy").reshape(-1,1)

display(b)
display(b.shape)
```

#### Lancia la DNN sul dataset considerato (compatibile SBNN)

```python
x = dataset
o = np.matmul(w.T, x.T)
display(o)
display(o.shape)

o = np.where(o < b, -1, 1)
display(o.T)
display(o.shape)
```
